**1. Copy**

**copy creates completely independent array with separate data.**

```python
import numpy as np
```

```python
a = np.array([10,20,30])

b = a.copy()

print("Original a: ", a)

print("copy array b: ",b)
b[0] = 99
print("After Change copy b: ",b)

# b has its own separate data, so changing b does not affect a.
```

**2. View**

**View creates a new array that shares the same data as the original array.**

```python
arr1 = np.array([10,20,30])

arr_b = arr1.view()

arr_b[0] = 100

print("Original: ", arr1)
print("View: ", arr_b)

# View = Same data, different array.
```

| Copy                          | View                        |
| ----------------------------- | --------------------------- |
| Creates separate data         | Shares original data        |
| Changes don't affect original | Changes can affect original |
| Uses more memory              | Uses less memory            |
| `a.copy()`                    | `a.view()`                  |


**3. Difference Between Copy and View**


```python
x = np.array([10,20,30])

copy_arr = x.copy()
view_arr = x.view()

print("Just Printing\n")
print("X: ", x)
print("Copy Arr: ",copy_arr)
print("View Arr: ",view_arr)

copy_arr[0] = 101
view_arr[0] = 999

print("\n After Change Print\n")

print("x: ", x)             # original are affted by view_arr 999
print("Copy Arr: ",copy_arr)
print("View Arr: ",view_arr)
```

**Why?**

**1. copy_arr** has separate data, so changing it doesn't affect x array.

**2. view_arr** shares data with x array, so changing it affects x array.


**Copy = Separate data**
**View = Shared data**

**4. Base**

**base tells you whether an array shares data with another array.**

```python
# Copy = No shared data
# View = Shared data

z = np.array([10,20,30])

m = z.view()
n = z.copy()

print(m.base)
print(n.base) # show none bcz copy has seperate data,  No shared data.
```

**5. Memory Sharing**

**Memory sharing means two arrays use the same underlying data in memory.**

```python
a = np.array([10,20,30])

b = a.view()

b[0] = 999

print(a)
print(b)

# memory sharing checking wheater two array actually shared the smae memory.
print(np.shares_memory(a,b))

c = a.copy()  # copy does not shared memory

print(np.shares_memory(a,c))  # False

# Here a and b share the same memory/data.
```
