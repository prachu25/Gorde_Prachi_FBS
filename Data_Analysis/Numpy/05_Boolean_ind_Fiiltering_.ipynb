```python
import numpy as np
```

**Comparison Operators**

**Comparison operators are used to compare array values and return True or False.**

```python
arr = np.array([232, 101, 450, 78, 320])

print(arr > 200)
print(arr < 200)
print(arr >= 200)
print(arr <= 200)
print(arr == 200)
print(arr != 200)
```

**Boolean Array**

**Is Array that Contains only True or False**

```python
res = arr > 200
print(res)
```

**Boolean Indexing**

**Using True or False Conditions to Select vallues from an Array**

```python
print(arr[arr > 200])
```

**Filtering Array**

**Filtering is selecting only the elements that satisfy a given condition.**

```python
arr1 = np.array([145, 67, 320, 89,210])
filter = arr1[arr1 > 100]
print("Print Values which Satisfiy Cond: ",filter) # 67 and 89 are smaller so thas why not print
```

**np.Where()**

**np.where() is used to find the index/position where a condition is true.**

```python
print(np.where(arr1 > 100))  # 145 320 120
```

**Multiple Conditions**

**Multiple conditions are used when we need to check two or more conditions on an array.(With NumPy, use & and |, not Python's and and or.)**

```python
new_arr = np.array([80,150,250,350,200])

result = new_arr[(new_arr > 100) & (new_arr < 300)]
print(result)

print(new_arr[(new_arr < 100) | (new_arr > 300)])

print(new_arr[~(new_arr > 100)])   # only 80 is less than 100
```

**np.logical_and()**

**np.logical_and() is used when both conditions must be True.**

```python
ans = np.logical_and((new_arr > 100), new_arr < 300)
print(ans)
```

**np.logical_or()**

**Is used when at least one condition should be true.**

```python
result = np.logical_or(new_arr < 100, new_arr > 300)
print(result)
print(new_arr[result])
```

**np.logical_not()**

**Is used to reverse a condition**
True -> False
False -> True

```python
res_not = np.logical_not(new_arr > 200)
print(res_not)
print(new_arr[res_not])
```

**2D Array Filtering**

**2D array filtering selects elements from a 2D array based on a condition.**

```python
arr = np.array([
    [120, 45, 300],
    [80, 250, 60]
])

# It checks every element and keeps only values greater than 100.
result = arr[arr > 100]

print(result)
```
