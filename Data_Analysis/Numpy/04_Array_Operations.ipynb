**What arithmetic operations does NumPy support?**

**NumPy supports element-wise arithmetic operations such as addition, subtraction, multiplication, division, modulus, and power.**

```python
import numpy as np

a = np.array([10,20,30])
b = np.array([2,4,5])
```

```python
print(a + b) # [12 24 35]
print(a - b) # [8 16 25]
print(a * b) # [20 40 150]
print(a / b)   # [5. 5. 6.]
print(a % b)   # [0 0 0]
print(a ** b)  # [100 160000 24300000]
```

**Mathematical Functions**

**NumPy provides mathematical functions such as sqrt(), abs(), round(), floor(), and ceil() for performing calculations on array elements.**

```python
c = np.array([10.4, 20.6, 30.7, 40.2])

d = np.array([-1.8,-67.8])


print("Sqrt: ",np.sqrt(c))  # square root

print("\nUsing abs: ",np.abs(d))   # absolute value- make all positive value

print(np.round(c)) # round

print("Floor: ",np.floor(c)) # round down
print("Round Up: ",np.ceil(c))  # round up
```

**Scaler Operation**

**A scalar operation is an operation performed between a NumPy array and a single value, applied element-wise to all elements.**

```python
arr = np.array([10,20,30,40])

print(arr)

print("Adding 5: ", arr + 5)
print("Sub 2: ", arr -2)
print("Mul 6: ", arr * 6)
print("Div 10: ", arr / 10)
```

**Comparison Opeartion**

**Comparison operations compare array elements and return a Boolean array containing True or False for each element.**

1. greater than >
2. less than <
3. greater than or equal >=
4. less than or equal <=
5. equal to ==
6. not equal to !=

```python
arr = np.array([10,20,30,40,50])

print(arr > 20) # 30 40 50
print(arr < 30) # 10 20
print(arr == 20) # 20
print(arr != 20) # 10   30 40 50
print(arr >= 20)
print(arr <= 10) # 10
```

**Univarsal Function**

**A ufunc, or Universal Function, is a NumPy function that performs operations element by element on array**

```python
arr = np.array([1, 4, 9, 16])
arr1 = np.array([-2,-5,4])
```

```python
print(np.sqrt(arr))   # absolute value
print(np.abs(arr1))   # square root
print(np.exp(arr))    # exponentail
print(np.log(arr))    # logarithm
print(np.sin(arr))    # sine
print(np.cos(arr))  # cosine
```

**Aggregation Operations**

**Aggregation means combining all array elements to get one result.**

```python
arr = np.array([10,20,30,40])
```

```python
print(np.sum(arr))  # sum of element in arr
print(np.prod(arr)) # product of all element in array
print(np.min(arr))  # min elem
print(np.max(arr))  # max ele
print(np.mean(arr)) # average
print(np.median(arr)) # median
print(np.std(arr)) # standard deviation
print(np.var(arr))        # variance
```
