**1. Random - np.random()**

**2. rand()**

np.random.rand() generates random numbers between 0 and 1.

```python
# one number
print(np.random.rand())
```

```python
# Multiple Random Numbers
print(np.random.rand(5))
```

```python
# 2D array
print(np.random.rand(2,3))   # 2 row , 3 col
```

**3. randint()**

randint() generates random integers within a specified range.

```python
# One Number
print(np.random.randint(1,10))
```

```python
# Generate 5 Number
print(np.random.randint(1,10,5))
```

```python
# 2D array
print(np.random.randint(1,100, size=(2,3)))
```

**4. randn()**

**np.random.randn() generates random numbers based on a standard normal distribution.**


```python
print(np.random.randn(5))  # it can create positive and negative values
```

```python
print(np.random.randn(2,3))
```

| `rand()`                             | `randn()`                          |
| ------------------------------------ | ---------------------------------- |
| Generates values between **0 and 1** | Generates values around **0**      |
| Uniform distribution                 | Standard normal distribution       |
| Values are always positive           | Values can be positive or negative |
| Example: `0.2, 0.7, 0.4`             | Example: `-1.2, 0.5, 1.3`          |


**5. random()**

**random() generates random floating-point numbers between 0 and 1**

```python
# One Number
print(np.random.random())
```

```python
# Multiple
print(np.random.random(4))
```

```python
# 2D array
print(np.random.random((2, 3)))
```

**6. Choice()**

np.random.choice() is used to randomly select values from an array.

```python
arr = np.array([10,20,30,40,50])

print(np.random.choice(arr))
# Each time we run it select diffrenet value
```

**7. shuffle()**

**np.random.shuffle() is used to randomly rearrange the elements of an array.**

```python
arr= np.array([10,20,30,40,50])

np.random.shuffle(arr)

print(arr)   # randomly reagrrange the original array
```

```python
r = np.random.shuffle(arr)
print(r)

# Why ? none
# shuffle change the original array directly and return None
```

**8.Permutation()**

**np.random.permutation() is used to randomly rearrange an array and return a new array.**

```python
array = np.array([10,20,30,40,50])

r = np.random.permutation(array)

print("Original: ", array)  # original not change

# permutation- does not change the original array, Return new shuffled array.
print("Permutation: ", r)
```

**1. shuffle**

**np.random.shuffle(arr)**


*   Change Original array
*   Return None



**2. permutataion()**

**r = np.random.permutation(arr)**



*   Does not change the original array
*   Return new shuffled array



**9. seed()**

**np.random.seed() is used to get the same random results every time.**

Without Seed()

```python
print(np.random.randint(1,10,5))
```

with seed()

```python
np.random.seed(10)

print(np.random.randint(1,10,5))

# Every time you run this code, we get the same sequence of random numbers.
```
