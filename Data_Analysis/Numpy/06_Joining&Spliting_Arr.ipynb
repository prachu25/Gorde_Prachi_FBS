**np.concatenate()**

**Is used to join two or more arrays together.**

**Statistics mein:**

1. axis=0 column-wise,

2. axis=1 row-wise.



**Concatenate mein:**

 1. axis=0 rows join,

 2. axis=1 columns join.

```python
import numpy as np

arr1 = np.array([10, 20, 30])
arr2 = np.array([40, 50, 60])
```

```python
res = np.concatenate((arr1, arr2))  # join two separate arr into one arr
print(res)
```

```python
arr2d1 = np.array([
    [10, 20],
    [30, 40]
])

arr2d2 = np.array([
    [50, 60],
    [70, 80]
])                             # axis = 0 row wise
                               # axis = 1 col wise
```

```python
np.concatenate((arr2d1, arr2d2))
# by deafult assus ekrta hai ki 0 so row wise
```

```python
np.concatenate((arr2d1,arr2d2), axis = 0) # 0 -> row wise
```

```python
np.concatenate((arr2d1,arr2d2), axis = 1) # 1 -> col wise
```

**np.stack()**

**np.stack() joins two or more arrays and creates a new axis.**

```python
print("arr1: ",arr1)
print("arr2: ",arr2)

print()

print("using Stack- it join arr1 & arr2 in 2D arr")
ans1 = np.stack((arr1,arr2))
print(ans1)  #
```

**np.vstack() - vertical stack**

**np.vstack() joins arrays vertically, one below another.**

```python
arr3 = np.array([70,80,90])

res = np.vstack((arr1,arr2,arr3))
print(res)

# So all three arrays are joined vertically, one below another.
```

**np.hstack() - horizaontal stack**

**np.hstack() joins arrays horizontally, side by side.**


```python
resh = np.hstack((arr1,arr2,arr3))
print(resh)
```

Diffrenece bwn vstack() vs hstack()

vstack() - one below another

hstack() - side by side

**# np.dstack()**

joins along 3rd axis

```python
array1 = np.array([
    [10, 20],
    [30, 40]
])

array2 = np.array([
    [50, 60],
    [70, 80]
])

result = np.dstack((array1, array2))

print(result)   # Here, the two 2D arrays are combined to create a 3D array.
```

**# np.split()**

**Is used to divide one array into multiple smaler arrays.**

[10, 20, 30, 40, 50, 60]

           ↓

[10, 20]  [30, 40]  [50, 60]

```python
split_arr = np.array([10,20,30,40,50,60])

ans2 = np.split(split_arr,3)  # 3 equal part

for i in ans2:
  print("Array: ",i)

print()

ans3 = np.split(split_arr,2)   # 2 equal part
for i in ans3:
  print("Array: ",i)
```

**np.array_split()**

**np.array_split() is similar to np.split(), but it can split an array into unequal parts.**

```python
unsplit = np.array([10,20,30,40,50])  # it Split UNEqual Part

un1 = np.array_split(unsplit,3)
print(un1)

print()

un2 = np.array_split(unsplit,4)
print(un2)
```

**np.vsplit()**

**Is used to split 2D array vertically, meaning it splits the array by rows**

```python
arr_1 = np.array([
    [10, 20],
    [30, 40],
    [50, 60],
    [70, 80]
])

result = np.vsplit(arr_1,2)

for i in result:
  print(i)
  print()

# vsplit() - split rows
# hsplit() - split columns
```

**np.hsplit()**

**np.hsplit() is used to split a 2D array horizontally, meaning it splits the array by columns.**

```python
result_col = np.hsplit(arr_1, 2)

for i in result_col:
  print(i)
  print()
```
