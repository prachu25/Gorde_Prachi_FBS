**What is np.sum() used for?**

**np.sum() is used to calculate the sum of all elements in a NumPy array.**

```python
# 1. np.sum()
import numpy as np
```

```python
# 1D
arr = np.array([10,20,30,40,50])
print(np.sum(arr))

# 2D
arr1 = np.array([
    [10,20],
    [5,99]
])

print(np.sum(arr1))
```

**What is np.min() used for?**

**np.min() is used to find the minimum value in a NumPy array.**

```python
print("Min Element 1D: ",np.min(arr))
print("Min Element 2D: ", np.min(arr1))
```

**What is np.max() used for?**

**np.max() is used to find the maximum value in a NumPy array.**

```python
print("Max Element 1D: ", np.max(arr))
print("Max Element 2D: ", np.max(arr1))
```

**What is np.mean() used for?**

**np.mean() is used to calculate the average value of elements in a NumPy array.**

```python
print("Mean from 1D: ",np.mean(arr))
print("Mean from 2D: ",np.mean(arr1))
```

**What is np.median() used for?**

**np.median() is used to find the middle value of an array after sorting its elements.**

```python
print("Median Element 1D: ",np.median(arr))  # internally sort but dont change the orignal array
print("Median Element 2D: ",np.median(arr1)) # internally sort but dont change the orignal array
```

**What is np.std() used for?**

**np.std() is used to measure how much the values in a NumPy array are spread out from their mean.**

```python
arr_s = np.array([10,20,30,40,50,60])
print(np.std(arr_s))
```

```python
# Small std - values are close to the mean
# Large std - values are more spread out
a = np.array([20, 21, 19, 20, 20])
b = np.array([5, 20, 35, 50, 65])

print(np.std(a))
print(np.std(b))
```

**what is np.var() used for?**

**np.var() is used to find how much the values are spread out from the average.**

```python
# 1D
array = np.array([10,20,30,40,50])

print("Variance: ", np.var(array))

# Relationship with std
# 1. variance = std²
# 2. std      = √variance
```

```python
# 2D
array2d = np.array([
    [10,20,30],
    [40,50,60]
])

print("2D variance: ", np.var(array2d))
```

**np.argmin() ?**

**np.argmin() is used to find the index of the smallest value in an array.**

```python
arr = np.array([232, 101, 450, 78, 320])
print("Min Value Index: ", np.argmin(arr)) # index = 3 , element = 78
```

**np.argax() ?**

**np.argmax() is used to find the index of the largest value in an array**

```python
print("Max Value index: ", np.argmax(arr)) #element = 450
```

```python
arr = np.array([
    [12, 85, 34],
    [67, 23, 91]
])

print(np.argmax(arr))  # element = 95
```

**Axis in numpy**

**Axis tells NumPy which direction to perform an operation.**

```python
arr = np.array([10, 20, 30])

print(np.sum(arr, axis=0))
```

```python
arr2dim = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

# axis=0 - uper se neeche, yani col pr operation
# 10  20  30
# ↓   ↓   ↓
# 40  50  60
# 50  70  90   # output
print("axis=0 COl pr operation: ",np.sum(arr2dim, axis = 0))   # axis =0


# Left se right, yani rows par operation.
print("axis=1 Row pr operation: ",np.sum(arr2dim, axis=1))

# 10 -> 20 -> 30  = 60
# 40 -> 50 -> 60  = 150

print()
print("axis=0 - coloums pr operation")
print("axis=1 - rows pr operation")
```

**Statistics with axis**

**Now we can use axis with statistical functions like sum, mean, min, max, std, and var.**

```python
# 1. np.sum() - Finds the total of all elements

arr = np.array([10, 20, 30, 40])

print(np.sum(arr)) # Output: 100


# 2. np.min() - Finds the smallest value
arr = np.array([232, 101, 450, 78, 320])
print(np.min(arr)) # Output: 78


# 3. np.max() - Finds the largest value
print(np.max(arr)) # Output: 450


# 4. np.mean() - Finds the average value
arr = np.array([10, 20, 30, 40])

print(np.mean(arr)) # Output: 25.0


# 5. np.median() - Finds the middle value

arr = np.array([50, 10, 30, 40, 20])

print(np.median(arr)) # Output: 30.0


# 6. np.std() - Finds how much values are spread out from the average

arr = np.array([10, 20, 30, 40, 50])

print(np.std(arr))


# 7. np.var() - Finds how much values are spread out from the average

arr = np.array([10, 20, 30, 40, 50])

print(np.var(arr)) # Output: 200.0


# 8. np.argmin() - Returns the index of the smallest value

arr = np.array([232, 101, 450, 78, 320])

print(np.argmin(arr)) # Output: 3


# 9. np.argmax() - Returns the index of the largest value

print(np.argmax(arr)) # Output: 2





# ==============================
# AXIS IN NUMPY
# ==============================

# Axis tells NumPy in which direction
# to perform an operation.


arr = np.array([
    [232, 101, 450],
    [78, 320, 190]
])


# axis=0 - column-wise
# axis=1 - row-wise


# SUM

print(np.sum(arr, axis=0))
print(np.sum(arr, axis=1))


# MIN

print(np.min(arr, axis=0))
print(np.min(arr, axis=1))


# MAX

print(np.max(arr, axis=0))
print(np.max(arr, axis=1))


# MEAN

print(np.mean(arr, axis=0))
print(np.mean(arr, axis=1))


# MEDIAN

print(np.median(arr, axis=0))
print(np.median(arr, axis=1))


# STD

print(np.std(arr, axis=0))
print(np.std(arr, axis=1))


# VAR

print(np.var(arr, axis=0))
print(np.var(arr, axis=1))


# ARGMIN

print(np.argmin(arr, axis=0))
print(np.argmin(arr, axis=1))


# ARGMAX

print(np.argmax(arr, axis=0))
print(np.argmax(arr, axis=1))


# ==============================
# IMPORTANT
# ==============================

# axis=0 - column-wise
# axis=1 - row-wise

# Statistical Functions:
# np.sum()
# np.min()
# np.max()
# np.mean()
# np.median()
# np.std()
# np.var()
# np.argmin()
# np.argmax()
```
