```python
import numpy as np
```

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(arr)
```

**1. ndim  → dimensions**

**2. shape → structure**

**3. size  → total elements**

**4. dtype → data type**

**1. ndim - Tell us how many dimensions the array has.**

```python
print(arr.ndim)
```

**2. shape - Tell us size of each dimension**

```python
print(arr.shape)  # o/p (2,3) i.e (2 row, 3 col)
```

```python
a = np.array([10,20,30,40])
print(a.shape)  # o/p = 4 . i.e 4 element in one dimension
```

**3. Size - Tell us total number of elements**

```python
print(arr.size)
```

**4. dtype - Tells us the data type stored inside the array.**

```python
print(arr.dtype)
```
