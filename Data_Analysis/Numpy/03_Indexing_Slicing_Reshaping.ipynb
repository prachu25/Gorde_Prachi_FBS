**What is Array Indexing?**

**Indexing means accessing an element from an array using its position.**

```python
import numpy as np
```

```python
arr = np.array([10,20,30,40])  # numpy arr index start from 0
print(arr[0])
```

**what is the Negative Index ?**

**Negative indexing is used to access array elements from the end using negative index values.**

```python
print(arr[-1])  # last element 40
print(arr[-2])  # second last element 30
```

**2D Indexing**

**for 2D array we use two indexes arr[row, col]**

```python
# 2D array indexing is used to access an element using its row and column index.
arr2d = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])

print(arr2d)
print(arr2d[1,1])  # print 50
print(arr2d[0,0])  # print 10
```

**Array Slicing**

arr[start : stop : step]

**Array slicing extracts a portion of an array using indexes.**

1. : - all / range
2. start: - start to end
3. :stop - beginning to stop
4. ::step - skip elements
5. [::-1] - reverse the array

```python
arr1 = np.array([90,60,80,30,20])

print(arr1[1:4])   # end not include
print(arr1[2:])
print(arr1[:3])
print(arr1[:])   # print arr

print(arr1[::2])  # start stop step
print(arr1[1::2])

print(arr1[::-1])  # reverse arr
print(arr1[3:0:-1]) # reverse from index 3
```

**2D array Slicing**

arr[row_slice, column_slice]

**Changing Element**

**We can modify NumPy array elements by assigning a new value using their index or slice.**

```python
# 1D Array

narr = np.array([10,20,30,40,50])
print(narr)
print("Element at index 3 = ",narr[3])
print()
narr[3] = 45
print("Element at index 3 After change = ", narr[3])
print(narr)

# Multiple Elements using scling  1:3 -> 101
print()
narr[1:4] = 101
print(narr)
```

```python
# 2D Array
narr2 = np.array([
    [10,20,30],
    [40,50,60]
])
print(narr2)

print()
narr2[1,2] = 100

print(narr2)
```

**Reshape()**

**reshape() changes the dimensions or shape of an array without changing its elements.**

```python
ar = np.array([10,20,30,40,50,60,70,80,90])

new_arr = ar.reshape(3,3)  # 3 row , 3 col
print(new_arr)

print()

ne_ar = ar.reshape(1,9)  # 1 row , 9  col
print(ne_ar)

print()

ne_ar1 = ar.reshape(9,1)  # 9 row, 1 col
print(ne_ar1)
```

**Flatten()**

**flatten() convert a 2D/ 3D array into 1D array**

**flatten() converts a multidimensional array into a 1D array and returns a copy**

```python
# 2D Array
arrf = np.array([
    [10,20,30],
    [40,50,60]
])
# 2D -> 1D
new_arr = arrf.flatten()  # convert it into 1d array
print(new_arr)

new_arr[0] = 101  #  not
print("\nFlatten Change or not\n")
print(new_arr)

print()
print(arrf)  # original array will not be change after change in new_aray
```

```python
# 3D array
arrf1 = np.array([
    [[10, 20], [30, 40]],
    [[50, 60], [70, 80]]
])

print(arrf1.shape)  # shape - size of each dimension

new_arr = arrf1.flatten()
print(new_arr)
```

**Ravel()**

**ravel() also convert multidimensional array into 1D**

```python
arr_r = np.array([
    [10,20,30],
    [40,50,60]
])

print(arr_r)
print()

new_arr = arr_r.ravel()
print(new_arr)

new_arr[0] = 101
print("\n new_array After Modifying\n")
print(new_arr)


print("\nOriginal Array\n")
print(arr_r)   # original Array also modify after chage in new_arr
print()
```

**flatten() vs ravel()**

**faltten() -**
1. return copy.
2. after change in new_arr the original array will not be affected.
3. use more memory.
4. sightly less memory- efficient.



**ravel() -**
1. Returns view when possible.
2. after change in new_arr the original array will also change.
3. uses less memory.
4. more memory efficient.

**Array Iteration**

**Array iteration means accessing each element of an array one by one using a loop.**

```python
# 1D array
arr = np.array([10,20,30,40])

for x in arr:
  print(x, end=" ")
```

```python
# 2D array
arr = np.array([
    [10,20],
    [30,40]
])

for row in arr:
  print(row)
  print()
```
