**1. Broadcasting Concept**

```python
import numpy as np
```

```python
arr = np.array([10,20,30])

# 5 is Added to every elemnet
print(arr + 5)

# Broadcasting = NumPy automatically adjusts smaller values/arrays
# to perform operations with larger arrays.
```

Rule- For broadcasting to work, the array shapes must be compatible.

```python
a = np.array([10, 20, 30]) # size = 3
b = np.array([1, 2, 3])    # size = 3

print(a + b) # both array have same shape so broadcasting Work.
```

```python
a = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

b = np.array([1, 2, 3])

print(a + b)
```

**2. Scalar BroadCasting**

**Scalar broadcasting means applying a single value to every element of an array during an operation.**

```python
s1 = np.array([10,20,30,40])
print(s1 * 5)
print(s1 + 2)
print(s1 - 2)
print(s1 /10)
```

**3. Array Broadcasting**

**Array Broadcasting: Performing operations between arrays of different shapes when their shapes are compatible.**

```python
a = np.array([10, 20, 30])
b = np.array([1])

print(a + b)
```

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a + b)
```

**4. Broadcasting Rules**

**Determine whether different-shaped arrays can perform operations together.**

**Main rules**
1. Dimensions are equal - works
2. One dimension is 1 - works
3. Missing dimensions are treated as 1
4. Otherwise - Broadcasting Error

**Rule 1:** If corresponding dimensions of two arrays are equal, broadcasting works.

```python
# For Rule 1: Same dimension = Compatible ✅
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a + b)
```

**Rule 2: If one corresponding dimension is 1, broadcasting works.**

```python
# for Rule 2: Dimension 1 = Compatible ✅
a = np.array([10, 20, 30])
b = np.array([5])

print(a + b)
```

**Rule 3:** If one array has fewer dimensions, NumPy treats the missing dimensions as 1.

```python
# for Rule 3: Missing dimension are Treated as 1
a = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

b = np.array([1, 2, 3])

print(a + b)

# a - (2, 3)  shape of a
# b - (3,)    shape of b
```

**Rule 4:** If the dimensions are not equal and neither dimension is 1, broadcasting does not work.

```python
# Rule 4: Incompatible Dimension
a = np.array([10, 20, 30, 40])
b = np.array([1, 2])

print(a + b) # then here brodcasting not work and show Error
```

**5.Shape Compatibility**

**Two arrays are shape-compatible when their dimensions are equal or one of them is 1.**

Equal or 1 = Compatible ✅

Different and neither is 1 = Not Compatible ❌

**6. 2D BroadCasting**

**A smaller array is automatically expanded to match a 2D array when the shapes are compatible.**

```python
a = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

b = np.array([1, 2, 3])

print(a + b)  # b is broadcast across each row.
```

**7. Broadcasting Errors**

**A broadcasting error occurs when two arrays have incompatible shapes.**

```python
a = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

b = np.array([1, 2])

print(a + b)   # Incompatible shapes = Broadcasting Error ❌
```

**8. Practical Broadcasting**

ex: real life example , add marks to every student

```python
marks = np.array([
    [70, 80, 90],
    [60, 75, 85]
])

bonus = np.array([5, 5, 5])

result = marks + bonus

print(result)   # bonus is automatically applied to each row.
```
