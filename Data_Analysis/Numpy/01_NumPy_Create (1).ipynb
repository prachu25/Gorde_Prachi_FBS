**NumPy - numeric Python**

```python
nums = [1,2,3,4,5]
sq = []

for i in nums:
  sq.append(i*i)

print(sq)
```

```python
nums = [1,2,3,4,5,6,7,8,9]

sq = [x*x for x in nums]  # using comprehension
print(sq)
```

**Creating Numpy array from 1D list**

```python
lst=[9,26,13,1,45,30,12,10,89,31]

arr = np.array(lst)

print(lst)
print(type(lst))


print(arr)
print(type(arr))
```

**Creating Numpy array from 2D list**

```python
lst = [
    [5,2,1,3],
    [4,7,8,9],
    [1,5,3,8]
]

print(lst)
```

```python
arr = np.array(lst)
print(arr)   # 2D numpy
```

**Creating numpy array with 0's and 1's**

```python
x = np.zeros(3)
print(x)   # 1D

print()

z = np.zeros((3,4))
print(z)   # 2D
```

```python
on = np.ones(4)
print(on)

print()
o = np.ones((4,5))
print(o)
```

```python
# we want to fill the another element use -> full((row,col),ele)
arr = np.full((3,3),5)
print(arr)

print("\n")

arr1 = np.full((4,3),8)
print(arr1)
```

**creating Identity Matrix**

```python
arr = np.eye(4)
print(arr)
```

**create numpy array using arange()**

```python
# creates a numbers within a range.
arr = np.arange(1,6)
print(arr)
```

```python
# np.arange(start, stop, step)
arr = np.arange(2,11,2)
print(arr)
```

```python
arr = np.arange(50,25,-3)
print(arr)
```

**create numpy array using arange()**

np.linspace() creates numbers with equal gaps between a start and end value.

```python
arr = np.linspace(1,10,2)
print(arr)
```

```python
arr = np.linspace(1,13,9)
print(arr)
```

**Multiple Types**

```python
lst = ['joe', 9.86, 45, True]

for i in lst:
  print(type(i))
```

```python
arr = np.array(lst)   # numpy stores the homogeneous data types only one type of data store
print(arr)            # it convert all into string

for i in arr:
  print(type(i))
```
