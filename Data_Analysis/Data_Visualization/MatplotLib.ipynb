### Bar Plot
1. Comparing Categories
2. Showing frequencies or count

```python
import matplotlib.pyplot as plt
```

```python
products = ['Hot coffee', 'Cold coffee', 'Tea', 'Salad']
summer = [30, 90, 80, 50]

plt.figure(figsize=(5, 3))  # width , height
plt.bar(products, summer, color="gray")

plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Summer Product Sales")

plt.show()
```
