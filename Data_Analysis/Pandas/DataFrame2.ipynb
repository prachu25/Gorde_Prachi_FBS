### Merge

```python
emp = pd.DataFrame({'id' :[101,102,103], 
                    'Name':['Ishan','Rohit', 'Hardik'], 
                    'did': [1,2,3]
                   })
emp
```

```python
dept = pd.DataFrame({'did' :[1,2,3,4,5], 
                     'dName':['IT','Sales', 'HR','Finance', 'Businees']
                    })
dept
```

```python
pd.merge(emp,dept)
```

```python
pd.merge(emp,dept, how = 'inner')
```

```python
pd.merge(emp,dept, how = 'outer')
```

```python
pd.merge(emp,dept, how = 'left')
```

```python
pd.merge(emp,dept, how = 'right')
```
