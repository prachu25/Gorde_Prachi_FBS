```python
import pandas
import numpy
```

**LOAD DATA**

```python
df = pandas.read_excel("netflix_DATA.xlsx")
print(df.head())
```

**null values check**

```python
df.isnull().sum()
```

```python
print(df.isnull().sum().sum())
```

```python
df.duplicated().sum()
```

**remove duplicates rows**

```python
# Remove Duplicates
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.shape
```

**rename column name**

```python
df.columns
```

```python
# rename columns
df.rename(columns={
    "Movie Title " : "movie_title",
    "Director" : "director",
    "Duration" : "duration",
    "Description" : "description"
}, inplace=True)
```

```python
df.columns
```

**Remove Unwanted colums**

```python
df.drop(columns=['cast_members','imdb_votes','box_office_usd','runtime_minutes','audience_gender'], inplace=True)
```

```python
df.drop('description', axis=1, inplace=True)
```

```python
df.columns
```

```python
print("\n SHAPE: ", df.shape)
```

```python
df.drop('rating', axis=1, inplace=True)
```

```python
print(df.shape)
```

**Unknown, N/A, NA Convert to NaN**

```python
df.isin(['Unknown', 'N/A', 'NA', '-']).sum()
```

```python
df.replace(['Unknown', 'N/A', 'NA', '-'], numpy.nan, inplace=True)
```

```python
df.isnull().sum()
```

```python
# Handel Movies null
df.dropna(subset=['movie_title'], inplace=True)
```

```python
df['movie_title'].isnull().sum()
```

```python
# types
df['type'].mode()
```

```python
print(df['type'].head(10))
```

```python
df['type'].isnull().sum()
```

```python
# fill missing value of type with unknown
df['type'] = df['type'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
# handel for director
df['director'] = df['director'].fillna('Not Specified')
```

```python
df.isnull().sum()
```

```python
# handel for country
df['country'] = df['country'].fillna('Unlisted')
```

```python
df.isnull().sum()
```

```python
# release_year data type check
df['release_year'].dtype
```

```python
# convert missing values to NaN
df['release_year'] = pandas.to_numeric(
    df['release_year'],
    errors='coerce'
)
```

```python
df['release_year'].isnull().sum()
```

```python
# drop missing year
df.dropna(subset=['release_year'], inplace=True)
```

```python
df['release_year'].isnull().sum()
```

```python
# Added DATE -> check the type
df['date_added'].dtype
```

```python
# convert it into the date
df['date_added'] = pandas.to_datetime(
    df['date_added'],
    format='mixed',
    errors='coerce'
)
```

```python
# check the type again
print(df['date_added'].dtype)
```

```python
df['date_added'] = df['date_added'].dt.strftime('%B %d, /%Y')
```

```python
df['date_added'].isnull().sum()
```

```python
# handel NaN
df['date_added'] = df['date_added'].fillna('Not Available')
```

```python
df['date_added'].isnull().sum()
```

```python
df['listed_in'].value_counts()
```

```python
df.isnull().sum()
```

```python
df['duration'].value_counts().head(10)
```

```python
df['duration'] = df['duration'].fillna('Not Avaliable')
```

```python
df.isnull().sum()
```

```python
df['imdb_rating'].dtypes
```

```python
df['imdb_rating'] = df['imdb_rating'].astype(float)
```

```python
df['imdb_rating'] = df['imdb_rating'].fillna(df['imdb_rating'].mean())
```

```python
df.isnull().sum()
```

```python
df['imdb_rating'].isnull().sum()
```

```python
df['language'] = df['language'].fillna('Not Specified')
```

```python
# check in language is thier missing or not
df['language'].isnull().sum()
```

```python
df['budget_usd'].dtype
```

```python
df['budget_usd'] = pandas.to_numeric(
    df['budget_usd'],
    errors='coerce'
)
```

```python
df['budget_usd'].dtype
```

```python
df['budget_usd'].isnull().sum()
```

```python
df.shape
```

```python
print(df.head(20))
```

```python
df['budget_usd'] = df['budget_usd'].fillna(df['budget_usd'].mean())
```

```python
df.duplicated().sum()
```

```python
# handel String
text_cols = [
    'movie_title',
    'type',
    'director',
    'country',
    'listed_in',
    'language'
]

# use strip to removes extra space
for col in text_cols:
  df[col] = df[col].str.strip()
```
