**DATA CLEANING**

```python
import pandas
```

**Load DATA**

```python
df = pandas.read_csv("rwa_data.csv")
```

```python
print(df.head(5))
```

```python
df.columns
```

```python
df.drop(['cast', 'imdb_votes'], axis=1, inplace=True)
```

```python
df.columns
```

```python
df.info()
```

```python
df.drop('description', axis=1, inplace=True)
```

```python
df.isnull().sum()
```

```python
df['director'].isnull().sum()
```

```python
# fill missing director values with unknown
df['director'].fillna('Unknown', inplace=True)
```

```python
df.isnull().sum()
```

```python
df['country'].isnull().sum()
```

```python
# fill missing country values with Not Specified
df['country'].fillna('Not Specified', inplace=True)
```

```python
df['country'].isnull().sum()
```

```python
df['language'].unique()
```

```python
df['language'] = df['language'].replace({

    # English
    'English': 'English',
    ' ENGLISH': 'English',
    'english ': 'English',
    'Eng': 'English',
    'English ': 'English',
    'english': 'English',

    # Spanish
    'Spanish': 'Spanish',
    'spanish': 'Spanish',
    ' SPANISH': 'Spanish',
    'Spanish ': 'Spanish',
    'Español': 'Spanish',

    # German
    'german': 'German',
    ' GERMAN': 'German',
    'German ': 'German',
    'German': 'German',

    # Hindi
    'hindi': 'Hindi',
    'Hind': 'Hindi',
    'Hindi': 'Hindi',
    ' Hindi': 'Hindi',
    ' HINDI': 'Hindi',
    'Hindi ': 'Hindi',

    # Korean
    'korean': 'Korean',
    ' KOREAN': 'Korean',
    'Korean': 'Korean',
    'Korean ': 'Korean',

    # French
    ' FRENCH': 'French',
    'French': 'French',
    'french': 'French',
    'French ': 'French',

    # Telugu
    'Telugu': 'Telugu',
    'telugu': 'Telugu',
    ' TELUGU': 'Telugu',
    'Telugu ': 'Telugu',

    # Kannada
    'Kannada ': 'Kannada',
    'Kannada': 'Kannada',
    ' KANNADA': 'Kannada',
    'kannada': 'Kannada'
})
```

```python
df['language'].unique()
```

```python
# fill missing language values
df['language'].fillna('Not Specified', inplace=True)
```

```python
df.isnull().sum()
```

```python
# Fill missing IMDb ratings with median
df['imdb_rating'].fillna(df['imdb_rating'].median(), inplace=True)
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
# Remove "min" from duration
df['duration'] = df['duration'].str.replace(' min', '', regex=False)
```

```python
#  convert duartion to numeric
df['duration'] = pandas.to_numeric(df['duration'], errors='coerce')
```

```python
df.dtypes
```

```python
df.isnull().sum()
```

```python
# fill missing duration with median
df['duration'].fillna(df['duration'].median(), inplace=True)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
# Remove  duplicate rows
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df.isnull().sum()
```

```python
df['date_added'] = df['date_added'].str.strip()
```

```python
df['date_added'] = pandas.to_datetime(
    df['date_added'],
    errors='coerce',
    format='mixed'
)
```

```python
df['date_added'].isnull().sum()
```

```python
df.dtypes
```
