```python
df['Technical_Test_Score'] = df['Technical_Test_Score'].fillna(df['Technical_Test_Score'].mean())
```

```python
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
df['Communication_Score'] = df['Communication_Score'].fillna(df['Communication_Score'].mean())
```

```python
print(df.isnull().sum()[df.isnull().sum() > 0])
```
