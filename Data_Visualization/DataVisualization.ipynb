```python
import pandas
import matplotlib.pyplot  as plt
```

**LINE Char-Growth & Decline**

```python
data = {
    "month":["Dec25","Jan26","Feb26","Mar26","Apr26","May26"],
    "sales":[20000,25000,17500,30000,21000,19700]
}

df = pandas.DataFrame(data)
print(df)
```

```python
plt.plot(df['month'], df['sales'], marker = "x", color = 'green')

plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()
plt.title("Sales in Last 6 months")
```

```python
# 1.first dataframe
df = pandas.DataFrame({
    'Year': [2020, 2021, 2022, 2023, 2024],
    'Revenue': [100, 150, 130, 180, 220]
})

plt.plot(df['Year'], df['Revenue'], color='gray')

# xticks tell dataframe to show my dataframe years
plt.xticks(df['Year'])

plt.xlabel("Year")
plt.ylabel("Revenue")
plt.title("Company Revenue")

plt.grid()
plt.show()
```

```python
# Line chart

plt.plot(df['Year'], df['Revenue'], marker='o', linestyle = '--', linewidth=2, markersize='8',color='blue')

plt.xticks(df['Year'])
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.title("Company Revenue")
plt.grid()
plt.show()
```

```python
df = pandas.DataFrame({
    'Year': [2020, 2021, 2022, 2023, 2024],
    'Revenue': [100, 150, 130, 180, 220],
    'Profit': [20, 30, 25, 40, 55]
})
```

```python
plt.plot(df['Year'], df['Revenue'], marker='o', label='Revenue')
plt.plot(df['Year'], df['Profit'], marker='o', label='Profit')

plt.xticks(df['Year'])
plt.xlabel("Year")
plt.ylabel("Amount")
plt.title("Company Revenue vs Profit")

plt.legend()
plt.grid()
plt.show()
```

**BAR CHARTS**

When do we use a Bar Chart?

Use a bar chart when you want to compare categories.

```python
# BAR CHARTS
df = pandas.DataFrame({
    'Genre': ['Action', 'Comedy', 'Drama', 'Horror'],
    'Movies': [120, 180, 150, 90]
})

plt.bar(df['Genre'], df['Movies'], color='indigo')
plt.xlabel("Genre")
plt.ylabel("Number of Movies")
plt.title("Movies By Genre")

plt.show()
```

```python
# Vertical
df = pandas.DataFrame({
    'Category': ['Electronics', 'Clothing', 'Books', 'Furniture'],
    'Sales': [500, 350, 200, 300]
})

plt.bar(df['Category'],
        df['Sales'],
        color ='royalblue'
        )
plt.xlabel("Categories")
plt.ylabel("Sales")
plt.title("Sales by Category")
plt.show()
```

```python
# horizontal

plt.figure(figsize=(6,4))
plt.barh(df['Category'],df['Sales'], color='teal')

plt.xlabel("Sales")
plt.ylabel("Category")
plt.title("Sales by Category")

plt.show()
```

**Histogram**

Used to understand how numerical values are distributed.

What does plt.hist() do?

It groups the ages into ranges (bins) and counts how many values fall into each range

```python
df = pandas.DataFrame({
    'Age': [18, 20, 21, 22, 22, 24, 25, 25, 26, 27,
            28, 29, 30, 30, 31, 32, 35, 36, 40, 45]
})
```

```python
plt.hist(df['Age'],bins=10, color='darkgreen', edgecolor='black')
# bins = number of groups/ranges

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distrubution")

plt.grid()
plt.show()
```

**Scatter Plot**

Shows relationship between two numerical variables

```python
df = pandas.DataFrame({
    'Experience': [1, 2, 3, 4, 5, 6, 7, 8],
    'Salary': [25, 30, 35, 42, 45, 52, 60, 68]
})
```

```python
plt.scatter(df['Experience'], df['Salary'], color='orange')

plt.xlabel("Exprience")
plt.ylabel("Salary")
plt.title("Exprience vs Salary")

plt.show()
```

```python
df = pandas.DataFrame({
    'Category': ['Electronics', 'Clothing', 'Books', 'Furniture'],
    'Sales': [500, 300, 150, 250]
})
```

```python
# autopct percentage display karta hai:
plt.pie(df['Sales'],labels=df['Category'],
        autopct='%1.1f%%',
        startangle=90
        )
plt.title("Sales Distrubution")
plt.show()
```

**BOX PLOT**

A Box Plot is a graph used to understand the distribution of numerical data and to identify outliers.

```python
df = pandas.DataFrame({
    'Salary': [25, 28, 30, 32, 35, 36, 38, 40, 42, 100]
})

plt.boxplot(df['Salary'])
plt.ylabel("Salary")
plt.title("Salary Distribution")
plt.show()
```

```python
df = pandas.DataFrame({
    'Java': [60, 65, 70, 72, 75, 80, 85],
    'Python': [55, 60, 65, 70, 72, 78, 95]
})

plt.boxplot([df['Java'], df['Python']], labels=['Java', 'Python'])

plt.ylabel("Marks")
plt.title("Java vs Python Marks")
plt.show()
```

**SubPlot**

Subplot is used to display multiple graphs in a single figure.

```python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

# subplot(rows, columns, position)
# means:

# 1 - 1 row
# 2 - 2 columns
# 1 - first position


plt.subplot(3, 2, 1)
plt.plot(x, y)
plt.title("Line")

plt.subplot(3, 2, 2)
plt.bar(x, y, color='green')
plt.title("Bar")

plt.subplot(3, 2, 3)
plt.scatter(x, y)
plt.title("Scatter")

plt.subplot(3, 2, 4)
plt.hist(y)
plt.title("Histogram")

plt.subplot(3, 2, 5)
plt.scatter(x, y)
plt.title("Scatter 2")

plt.subplot(3, 2, 6)
plt.hist(y)
plt.title("Histogram 2")

plt.tight_layout()
plt.show()
```

**Chart customization**

means changing the size, labels, title, colors, markers, grid, and other properties of a chart to make it clear and readable.

```python
df = pandas.DataFrame({
    'Category': ['Electronics', 'Clothing', 'Books', 'Furniture'],
    'Sales': [500, 350, 200, 300]
})

plt.figure(figsize=(10, 6))

bars = plt.bar(
    df['Category'],
    df['Sales'],
    color='teal'
)

plt.bar_label(bars)
plt.xlabel("Category")
plt.ylabel("Sales")
plt.title("Category Wise Sales")

plt.xticks(rotation=30)

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

**Pandas .plot()**

is used to create different types of charts directly from a DataFrame.

```python
df = pandas.DataFrame({
    'Category': ['Electronics', 'Clothing', 'Electronics', 'Books', 'Clothing'],
    'Sales': [500, 300, 700, 200, 400]
})

df.plot(
    x='Category',
    y='Sales',
    kind='bar',
    color='salmon'
)
plt.xticks(rotation=30)
plt.title("Product Wise Sales")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

**GroupBy + Visualization - Multiple Aggregations**

```python
df = pandas.DataFrame({
    'Category': ['Electronics', 'Clothing', 'Electronics', 'Books', 'Clothing'],
    'Sales': [500, 300, 700, 200, 400],
    'Profit': [100, 50, 150, 40, 80]
})
```
